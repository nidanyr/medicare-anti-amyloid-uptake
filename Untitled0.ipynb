```python
import duckdb

query = """
SELECT
    HCPCS_Cd                        AS hcpcs_code,
    Brnd_Name                       AS brand_name,
    Gnrc_Name                       AS generic_name,
    CAST(yr AS INTEGER)             AS year,
    tot_spending,
    tot_claims,
    tot_benes,
    CASE
        WHEN tot_spending IS NULL AND tot_claims IS NULL AND tot_benes IS NULL
        THEN 'Not marketed / no Part B claims'
        ELSE 'Reported'
    END                             AS data_status
FROM partb
UNPIVOT INCLUDE NULLS (
    (tot_spending, tot_claims, tot_benes)
    FOR yr IN (
        (Tot_Spndng_2020, Tot_Clms_2020, Tot_Benes_2020) AS '2020',
        (Tot_Spndng_2021, Tot_Clms_2021, Tot_Benes_2021) AS '2021',
        (Tot_Spndng_2022, Tot_Clms_2022, Tot_Benes_2022) AS '2022',
        (Tot_Spndng_2023, Tot_Clms_2023, Tot_Benes_2023) AS '2023',
        (Tot_Spndng_2024, Tot_Clms_2024, Tot_Benes_2024) AS '2024'
    )
)
WHERE TRIM(HCPCS_Cd) IN ('J0172', 'J0174', 'J0175')
ORDER BY hcpcs_code, year
"""

df = con.sql(query).df()
df
```
